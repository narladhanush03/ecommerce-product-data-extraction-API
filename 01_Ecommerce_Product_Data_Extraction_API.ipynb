# E-commerce Product Data Extraction & Processing Using REST API

## Objective

Extract e-commerce product data from a REST API, process and clean the JSON response using Python and Pandas, validate the dataset, create useful derived fields, and export the final dataset as a CSV file ready for data analysis.

#### 1. Importing Libraries

```python
import requests
import pandas as pd
```

#### 2. API Connection

```python
url = "https://dummyjson.com/products"
response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
response.raise_for_status()
```

#### 3. Data Extraction

```python
params = {"limit": 0}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

print("Total Products:", data["total"])
print("Products Received:", len(data["products"]))
```

#### 4. JSON → Pandas DataFrame

```python
df = pd.DataFrame(data["products"])

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df.head()
```

```python
df.info()
```

#### 5. Missing values

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

#### 6. Selecting Analysis Columns

```python
analysis_cols = [
    "id",
    "title",
    "category",
    "price",
    "discountPercentage",
    "rating",
    "stock",
    "brand",
    "sku",
    "weight",
    "availabilityStatus",
    "returnPolicy",
    "minimumOrderQuantity"
]

df_analysis = df[analysis_cols].copy()

print("Analysis Dataset Shape:", df_analysis.shape)
```

```python
df_analysis.head()
```

### 7. Data Cleaning

#### 7.1. Handling Missing Values

```python
df_analysis["brand"] = df_analysis["brand"].fillna("Unknown")

print("Missing brand values:", df_analysis["brand"].isnull().sum())
```

#### 7.2. Duplicate Product IDs

```python
duplicate_ids = df_analysis["id"].duplicated().sum()

print("Duplicate Product IDs:", duplicate_ids)
```

#### 7.3. Duplicate SKUs

```python
duplicate_skus = df_analysis["sku"].duplicated().sum()

print("Duplicate SKUs:", duplicate_skus)
```

### 8. Data Validation

#### 8.1. Negative Prices

```python
negative_prices = (df_analysis["price"] < 0).sum()

print("Negative Prices:", negative_prices)
```

#### 8.2. Negative Stock

```python
negative_stock = (df_analysis["stock"] < 0).sum()

print("Negative Stock:", negative_stock)
```

#### 8.3 Invalid Discounts

```python
invalid_discounts = (
    (df_analysis["discountPercentage"] < 0) |
    (df_analysis["discountPercentage"] > 100)
).sum()

print("Invalid Discounts:", invalid_discounts)
```

#### 8.4. Invalid Ratings

```python
invalid_ratings = (
    (df_analysis["rating"] < 0) |
    (df_analysis["rating"] > 5)
).sum()

print("Invalid Ratings:", invalid_ratings)
```

#### 8.5. Invalid Minimum Order Quantities

```python
invalid_min_qty = (
    df_analysis["minimumOrderQuantity"] <= 0
).sum()

print("Invalid Minimum Order Quantities:", invalid_min_qty)
```

### 9. Feature Engineering

#### 9.1. Discount Amount

```python
df_analysis["Discount_Amount"] = (
    df_analysis["price"] *
    df_analysis["discountPercentage"] / 100
).round(2)
```

#### 9.2. Discounted Price

```python
df_analysis["Discounted_Price"] = (
    df_analysis["price"] -
    df_analysis["Discount_Amount"]
).round(2)
```

#### 9.3. Inventory Value

```python
df_analysis["Inventory_Value"] = (
    df_analysis["price"] *
    df_analysis["stock"]
).round(2)
```

#### 9.4. Discounted Inventory Value

```python
df_analysis["Discounted_Inventory_Value"] = (
    df_analysis["Discounted_Price"] *
    df_analysis["stock"]
).round(2)
```

#### 9.5. Discount Band

```python
def discount_band(discount):
    if discount == 0:
        return "No Discount"
    elif discount <= 10:
        return "Low"
    elif discount <= 25:
        return "Medium"
    else:
        return "High"

df_analysis["Discount_Band"] = (
    df_analysis["discountPercentage"].apply(discount_band)
)
```

#### 9.6. Stock Status

```python
def stock_status(stock):
    if stock == 0:
        return "Out of Stock"
    elif stock <= 20:
        return "Low Stock"
    elif stock <= 50:
        return "Medium Stock"
    else:
        return "High Stock"

df_analysis["Stock_Status"] = (
    df_analysis["stock"].apply(stock_status)
)
```

#### 10. Final Validation

```python
print("Final Dataset Shape:", df_analysis.shape)

print("\nMissing Values:")
print(df_analysis.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df_analysis.duplicated().sum())
```
